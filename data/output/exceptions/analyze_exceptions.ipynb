# Exception Analysis

This notebook analyzes data quality exceptions from the pipeline.

```python
import pandas as pd
from pathlib import Path

EXCEPTIONS_DIR = Path('.')
```

```python
exception_files = list(EXCEPTIONS_DIR.glob('*.csv'))
print(f"Found {len(exception_files)} exception files:")
for f in sorted(exception_files):
    df = pd.read_csv(f)
    print(f"  {f.name}: {len(df):,} rows")
```

## Address Fill Exceptions

```python
if (EXCEPTIONS_DIR / '02-fill-addresses-still-missing.csv').exists():
    df = pd.read_csv(EXCEPTIONS_DIR / '02-fill-addresses-still-missing.csv')
    print(f"Parcels still missing addresses: {len(df):,}")
    print("\nBy neighborhood:")
    print(df['analysis_neighborhood'].value_counts().head(10))
else:
    print("No address exceptions found")
```

## Zoning Fill Exceptions

```python
if (EXCEPTIONS_DIR / '10-fill-zoning-imputed.csv').exists():
    df = pd.read_csv(EXCEPTIONS_DIR / '10-fill-zoning-imputed.csv')
    print(f"Parcels with zoning imputed via spatial join: {len(df):,}")
    print("\nBy neighborhood:")
    print(df['analysis_neighborhood'].value_counts().head(10))
    print("\nBy imputed zoning code:")
    print(df['FZP Planning Code'].value_counts().head(10))
else:
    print("No zoning imputation exceptions found")
```

```python
if (EXCEPTIONS_DIR / '10-fill-zoning-still-missing.csv').exists():
    df = pd.read_csv(EXCEPTIONS_DIR / '10-fill-zoning-still-missing.csv')
    print(f"Parcels still missing zoning: {len(df):,}")
    print("\nBy neighborhood:")
    print(df['analysis_neighborhood'].value_counts().head(10))
else:
    print("No zoning still-missing exceptions found")
```

## Height Fill Exceptions

```python
if (EXCEPTIONS_DIR / '12-fill-height-imputed.csv').exists():
    df = pd.read_csv(EXCEPTIONS_DIR / '12-fill-height-imputed.csv')
    print(f"Parcels with height imputed via spatial join: {len(df):,}")
    print("\nBy neighborhood:")
    print(df['analysis_neighborhood'].value_counts().head(10))
    print("\nHeight distribution:")
    print(df['Height_Ft'].describe())
else:
    print("No height imputation exceptions found")
```

```python
if (EXCEPTIONS_DIR / '12-fill-height-still-missing.csv').exists():
    df = pd.read_csv(EXCEPTIONS_DIR / '12-fill-height-still-missing.csv')
    print(f"Parcels still missing height: {len(df):,}")
    print("\nBy neighborhood:")
    print(df['analysis_neighborhood'].value_counts().head(10))
else:
    print("No height still-missing exceptions found")
```

## SDB Computed Exceptions

```python
if (EXCEPTIONS_DIR / '17-fill-sdb-computed.csv').exists():
    df = pd.read_csv(EXCEPTIONS_DIR / '17-fill-sdb-computed.csv')
    print(f"Parcels with SDB computed (not from FZP data): {len(df):,}")
    print("\nBy neighborhood:")
    print(df['analysis_neighborhood'].value_counts().head(10))
    print("\nSDB qualification breakdown:")
    print(df['SDB_2016_5Plus'].value_counts())
else:
    print("No SDB computed exceptions found")
```

```python
if (EXCEPTIONS_DIR / '18-fill-historic-computed.csv').exists():
    df = pd.read_csv(EXCEPTIONS_DIR / '18-fill-historic-computed.csv')
    print(f"Parcels with historic status computed (not from FZP data): {len(df):,}")
    print("\nBy neighborhood:")
    print(df['analysis_neighborhood'].value_counts().head(10))
    print("\nHistoric status breakdown:")
    print(df['Historic'].value_counts())
else:
    print("No historic computed exceptions found")
```
